#### isin() function

```python
# find the most deadliest player
# highest strike rate = (number of runs made/ number of balls played)* 100
# in death over (16-20) and the player must have played > 200 balls in death overs in lifetime --> consider only such players
```

```python
death_over = delivery[delivery['over'] > 15]
```

```python
all_batsman = death_over.groupby('batsman')['batsman_runs'].count()
x = all_batsman > 200
batsman_list = all_batsman[x].index.tolist()
```

```python
# now we need runs and balls played by these batsman only
# we can't create mask for all 43 players so we use isin() function --> isin takes in only lists and returns a boolean series
```

```python
final = delivery[delivery['batsman'].isin(batsman_list)]
```

```python
runs = final.groupby('batsman')['batsman_runs'].sum()
```

```python
balls = final.groupby('batsman')['batsman_runs'].count()
```

```python
sr = ((runs/balls)*100).sort_values(ascending = False)
```

## Handling missing data

#### Drop method: dropna()

```python
foodie = food.pivot_table(index = ['City', 'Gender'], columns = 'Frequency', values = 'Spends', aggfunc = 'mean')
```

```python
foodie.head()
```

```python
foodie.shape
```

```python
foodie.dropna().shape # it has deleted all the rows with NaN (no values)
# since how attribute has default value as 'any', it goes up to evry row and deletes that row if any of its column has NaN
# that is, none of 12 rows had all its column filled with NaN
```

```python
foodie.dropna(how = 'all').shape # gives those rows which has NaN in ateast one of its columns -> all rows have NaN atleast once
```

```python
# this applied on rows above, to apply same on columns, then axis = 1 will be attribute while axis has its default value as 0 for rows
```

```python
foodie.dropna(subset = ['Daily']).shape # drops 8 rows with NaN in 'Daily' column
```

```python
foodie.dropna(subset = ['Daily', 'Never']).shape # now deletes those rows which has NaN in both Daily and Never (& function)
```

#### Fill: fillna()

```python
foodie.fillna(0) # fills all the data with NaN as 0, including the string data too which is problematic
foodie['Daily'].fillna('Not specified') # to make it permanent, inplace = True --> applies on foodie data itself even
# when the changes are made on column
```
