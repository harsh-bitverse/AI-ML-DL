```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#### 2D Line plot - for bivariate columns
use case - Time series data

```python
price = [48000, 54000, 57000, 49000, 47000, 45000]
year = [2015, 2016, 2017, 2018, 2019, 2020]
plt.plot(year, price) # if two columns are numerical and categorical respectively --> categorical data should be on X axis
```

```python
batsman = pd.read_csv('/content/sharma-kohli.csv')
batsman
```

```python
plt.style.use('classic')
plt.plot(batsman['index'], batsman['V Kohli'])
```

```python
plt.style.use('default')
plt.plot(batsman['index'], batsman['V Kohli'], color = 'green', linestyle = 'dashed', linewidth = 3, marker = 'o', markersize = 5, label = 'Virat')
# default linestyle = 'solid'
plt.plot(batsman['index'], batsman['RG Sharma'], color = '#000000', linestyle = 'dashdot', marker = 'D', label = 'Rohit') # #000000 is hex code for daark black
# markers are the points against which the graph is plotted, 'D' for Diamond, '+' for plus, '.' for dot, '>' for arrows, 'o' for bigger dot
plt.title('VK vs RS runs comparison')
plt.xlabel('Season')
plt.ylabel('Runs scored')
plt.legend(loc = 'upper right') # must call this function to label the lines plotted using label = 'Virat' and 'Rohit' --> loc = 'best' is default which sits in the space
# where there is free space
plt.grid() # to apply the grid
```

```python
price = [48000, 54000, 57000, 49000, 47000, 45000, 4500000] # there is data outlier here due to which the variation among previous ones is neglected
# so we use ylim
year = [2015, 2016, 2017, 2018, 2019, 2020, 2021]
plt.plot(year, price) # if two columns are numerical and categorical respectively --> categorical data should be on X axis
plt.ylim(0, 100000) # sets the range to which the data on y be limited to --> used to neglect the outlier to see the actual variation
plt.xlim(2017, 2019) # includes both 2017 and 2019
```

#### Bar Chart : Univariate/ Bivariate
use case : aggregate analysis of the groups --> Numerical vs Categorical data

```python
children = [10, 20, 40, 10, 30]
color = ['Red', 'Blue', 'Green', 'Yellow', 'Pink']
plt.bar(color, children, color = 'Black') # can apply all the attributes studied till now
```

```python
children = [10, 20, 40, 10, 30]
color = ['Red', 'Blue', 'Green', 'Yellow', 'Pink']
plt.barh(color, children, color = 'Black') # for horizontal bar chart
```

```python
df = pd.read_csv('/content/batsman_season_record.csv')
df
```

```python
# we don't have direct method to produce multi bars consecutively, so we do as follows
plt.bar(np.arange(df.shape[0]) - 0.2, df['2015'], width = 0.2)
plt.bar(np.arange(df.shape[0]), df['2016'], width = 0.2)
plt.bar(np.arange(df.shape[0]) + 0.2, df['2017'], width = 0.2)

# to replace x axis numbers with batsman names
plt.xticks(np.arange(df.shape[0]), df['batsman']) # first give the one which needs to be changed
plt.show()
```

```python
children = [10, 20, 40, 10, 30]
color = ['Red Red Red Red Red', 'Blue Blue Blue Blue Blue', 'Green Green Green Green Green', 'Yellow Yellow Yellow Yellow Yellow', 'Pink Pink Pink Pink Pink']
# when the names of the categorical data is huge, then we use xticks as follows to avoid their overlapping
plt.bar(color, children, color = 'Black')
plt.xticks(rotation = 'vertical')
```

#### Pi Chart : Univariate/ Bivariate (categorical vs numerical)
use case : to find contribution on standard scale

```python
import matplotlib.pyplot as plt
import pandas as pd
data = [23, 45, 100, 20, 49]
subjects = ['eng', 'kan', 'sc', 'ssc', 'maths']
plt.pie(data, labels = subjects) # it is labels in here --> not label
plt.show()
```

```python
df = pd.read_csv('/content/gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'], labels = df['batsman'], autopct = '%0.1f%%', colors = ['black', 'green', 'pink', 'yellow'])
# here 0.1f% represents percentage upto 1 decimal
# you can change 0.1 of even f to g, i, c, etc giving different representations of same percentage
# if less number of colors wrt number of divisions on pie are given then when the colors are over, it again starts from beginning
# and picks black and green for next 2 players as shown
plt.show()
```

```python
plt.pie(df['batsman_runs'], labels = df['batsman'], autopct = '%0.1f%%', explode = [0.1, 0, 0, 0, 0, 0], shadow = True)
# explode - to separate a piece from others --> on changing the values from 0.1 to any other, the piece will be separated with
# that much of distance
plt.show()
```
